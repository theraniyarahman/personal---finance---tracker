```python
print("PERSONAL FINANCE TRACKER")

#initialize total income and total expenses to 0
total_income=0
total_expense=0

#loop to calculate total income
choice='yes'
while choice=='yes':
    income= float(input("Enter income: "))
    total_income+= income

##input validation loop
    while True:
      choice=input("Add more income? type 'yes' or 'no': ").lower()
      if choice in ['yes','no']:
        break #valid input
      print("Invalid input! please enter 'yes' or 'no'") #invalid input


print("-----------------------------------------")

#loop to calculate total expenses
choice='yes'
while choice=='yes':
    expense= float(input("Enter Expense: "))
    total_expense+= expense

#input validation loop
    while True:
      choice=input("Add more expense? type 'yes' or 'no': ").lower()
      if choice in ['yes','no']:
        break #valid input
      print("Invalid input! please enter 'yes' or 'no'") #invalid input


# to calculate total savings
net_savings=total_income-total_expense

#display summary and financial status
print("-----------------------------------------")
print("FINAL SUMMARY REPORT")
print(f"Total Income  : ₹{total_income}")
print(f"Total Expense : ₹{total_expense}")
print(f"Net Savings   : ₹{net_savings}")

print("-----------------------------------------")
print("FINANCIAL STATUS:")
if (net_savings>0):
  print("You are saving money. Good Job!")
elif (net_savings==0):
   print("No savings.")
else:
   print("Oops! You are in Loss.")
```
